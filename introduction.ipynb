# 🧭 Fair dice analyses

An interactive collection for comparing raw randomness with methods that transform biased physical sources into uniform outcomes. Each notebook can be run independently and explains the method, simulates representative bias levels, and visualizes the result.

## Analyses

- 🪙 [Coin experiment](coin_experiment.ipynb) — tests the two-toss highest-slot method on fair and biased coins.
- 🎲 [Six-face dice experiment](dice_6_faces_experiment.ipynb) — compares raw six-sided rolls with extractor-corrected uniform slots.
- 0️⃣1️⃣ [Six-face direct binary mapping](dice_6_faces_binary_experiment.ipynb) — uses one roll per bit: faces 1–3 map to 0 and 4–6 map to 1.
- 🎲 [Eight-face dice experiment](dice_8_faces_experiment.ipynb) — measures correction quality and acceptance cost for eight-sided dice.
- 🎲 [Sixteen-face dice experiment](dice_16_faces_experiment.ipynb) — scales the same extraction idea to sixteen output positions.
- 📊 [Fair dice experiment](fair_dice_experiment.ipynb) — an expanded comparison of raw bias, corrected results, and efficiency.
- ⚖️ [Extractor efficiency comparison](extractor_efficiency_comparison.ipynb) — compares the expected tosses/rolls needed for a valid result across every extractor and bias profile.
- 🔐 [Binary entropy files](binary_entropy_files.ipynb) — writes secure random data and a raw SHA-256 chain, then checks the random file for collisions, bit balance, byte frequencies, and Shannon entropy.
